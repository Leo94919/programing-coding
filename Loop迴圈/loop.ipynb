# Loop

Loop指令可以讓程式碼依我的要求不斷執行，直到條件滿足為止

```python
# 底下的程式碼會印出0-9
for i in range(10):
    print(i)
```

```python
list(range(10))
```

```python
# 底下的程式碼會印出0-9
for i in range(10):
    print(i+1)
```

```python
sum=0
for i in range(10):
    sum=sum+(i+1)
print(sum)
```

# 九九乘法表

1\*=1  1\*2=2

```python
i=1
for j in range(1,6):
    print(f'{i}*{j}={i*j}', end=',')
print()
i=2
for j in range(1,6):
    print(f'{i}*{j}={i*j}', end=',')
```

```python
for i in range(1,6):
    for j in range(1,6):
        print(f'{i}*{j}={i*j}', end=',')
    print()
```
